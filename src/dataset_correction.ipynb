```python
import json

# Load jsonl file into a list of dictionaries
with open("../data/labelled/all.jsonl") as test_f:
    data = [json.loads(line) for line in test_f]
```

```python
for item in data:
    last_label, last_entity = "", ""
    for ent in item["label"]:
        start, end, tag = ent
        delimiter, label = tag.split("-")

        if delimiter == "I" and last_label != label:
            print(
                f"mismatch detected: {ent}; last entity: {last_entity}, id: {item['id']}"
            )
            ent[2] = f"B-{label}"
            # print(ent[2])
        # print(delimiter, label)

        last_entity = ent
        last_label = label
```

```python
# Save to JSON
with open("../data/labelled/all_corrected.jsonl", "w") as output_file:
    json.dump(data, output_file)
```

```python
from util import combine_bio_entities, replace_with_parentheses

for item in data:
    text = item["text"]
    labels = combine_bio_entities(item["label"])

    pred = replace_with_parentheses(text, labels)
    item["input"] = text
    item["text"] = pred
```

```python
with open("../data/labelled/all_parallel_corrected.jsonl", "w") as outfile:
    for entry in data:
        json.dump(entry, outfile)
        outfile.write("\n")
```
