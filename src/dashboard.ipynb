```python
import pandas as pd
import numpy as np

results = pd.read_json("./experiments_results.json")
# results = results.rename(
#     {
#         "eval_precision": "precision",
#         "eval_recall": "recall",
#         "eval_f1": "f1",
#     }
# )
results = results.T
results["precision"] = results["precision"].replace(np.nan, 0) + results[
    "eval_precision"
].replace(np.nan, 0)
results["recall"] = results["recall"].replace(np.nan, 0) + results[
    "eval_recall"
].replace(np.nan, 0)
results["f1"] = results["f1"].replace(np.nan, 0) + results["eval_f1"].replace(np.nan, 0)

results = results.sort_index(ascending=False)
results = results[["precision", "recall", "f1"]]
```

```python
import matplotlib.pyplot as plt
import numpy as np

results_baseline = results.loc[results.index.str.contains("bert", case=False)]
results_baseline[["precision", "recall", "f1"]].plot(
    kind="barh",
    figsize=(10, 7),
)
plt.grid()
plt.xticks(np.arange(0, 1.1, 0.05))
plt.title("Baseline results")
plt.xlabel("Metrics score")
plt.ylabel("Model")

print()
```

```python
import json

p = "/home/giuseppetermerissa/Scrivania/tesi-gtermerissa/output/rener/rener_2025-11-10_13-29-12_llama70b.json"

with open(p) as res_file:
    res = json.load(res_file)

precision_70b = []
recall_70b = []
f1_70b = []
for k in range(0, 7):
    precision_70b.append((res[f"top_k={k}"]["precision"]))
    recall_70b.append((res[f"top_k={k}"]["recall"]))
    f1_70b.append((res[f"top_k={k}"]["f1"]))
```

```python
import matplotlib.pyplot as plt
import numpy as np

precision_70b = []
recall_70b = []
f1_70b = []

for k in range(0, 7):
    precision_70b.append(res[f"top_k={k}"]["precision"])
    recall_70b.append(res[f"top_k={k}"]["recall"])
    f1_70b.append(res[f"top_k={k}"]["f1"])

# Set up the plot
x = np.arange(len(precision_70b))  # the label locations
width = 0.25  # the width of the bars

fig, ax = plt.subplots(figsize=(10, 6))

# Create bars
bars1 = ax.bar(x - width, precision_70b, width, label="Precision")
bars2 = ax.bar(x, recall_70b, width, label="Recall")
bars3 = ax.bar(x + width, f1_70b, width, label="F1")

# Add labels and title
ax.set_xlabel("k", fontsize=12)
ax.set_ylabel("Score", fontsize=12)
ax.set_title("Precision, Recall, and F1 Score by k value - Llama3.1 70B", fontsize=14)
ax.set_xticks(x)
ax.set_xticklabels(range(0, 7))
ax.legend()
ax.grid(axis="y", alpha=0.3)

# Set y-axis limits based on data range
all_values = precision_70b + recall_70b + f1_70b
y_min = min(all_values) - 0.05
y_max = max(all_values) + 0.005
ax.set_ylim(y_min, y_max)


# Add value labels on bars (optional)
def autolabel(bars):
    for bar in bars:
        height = bar.get_height()
        ax.annotate(
            f"{height:.2f}",
            xy=(bar.get_x() + bar.get_width() / 2, height),
            xytext=(0, 3),  # 3 points vertical offset
            textcoords="offset points",
            ha="center",
            va="bottom",
            fontsize=8,
        )


autolabel(bars1)
autolabel(bars2)
autolabel(bars3)

plt.tight_layout()
plt.show()
```
