```python
import pandas as pd
import numpy as np

results = pd.read_json("../output/experiments_results.json")
# results = results.rename(
#     {
#         "eval_precision": "precision",
#         "eval_recall": "recall",
#         "eval_f1": "f1",
#     }
# )
results = results.T
results["precision"] = results["precision"].replace(np.nan, 0) + results[
    "eval_precision"
].replace(np.nan, 0)
results["recall"] = results["recall"].replace(np.nan, 0) + results[
    "eval_recall"
].replace(np.nan, 0)
results["f1"] = results["f1"].replace(np.nan, 0) + results["eval_f1"].replace(np.nan, 0)

results = results.sort_values(by="precision", ascending=False)
results = results[["precision", "recall", "f1"]]
```

```python
import matplotlib.pyplot as plt
import numpy as np

results[["precision", "recall", "f1"]].plot(
    kind="barh",
    figsize=(10, 7),
)
plt.grid()
plt.xticks(np.arange(0, 1.1, 0.1))

print()
```
